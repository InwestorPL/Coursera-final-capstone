```python
# Task 2: Mark the success/failed launches for each site on the map
```

Next, let's try to enhance the map by adding the launch outcomes for each site, and see which sites have high success rates.
Recall that data frame spacex_df has detailed launch records, and the `class` column indicates if this launch was successful or not


```python
spacex_df.tail(10)
```

Next, let's create markers for all launch records.
If a launch was successful `(class=1)`, then we use a green marker and if a launch was failed, we use a red marker `(class=0)`


Note that a launch only happens in one of the four launch sites, which means many launch records will have the exact same coordinate. Marker clusters can be a good way to simplify a map containing many markers having the same coordinate.


Let's first create a `MarkerCluster` object


```python
marker_cluster = MarkerCluster()
```

*TODO:* Create a new column in `spacex_df` dataframe called `marker_color` to store the marker colors based on the `class` value


```python

# Apply a function to check the value of `class` column
# If class=1, marker_color value will be green
# If class=0, marker_color value will be red
# 1. Definimos uma função auxiliar para decidir a cor
def assign_marker_color(launch_class):
    if launch_class == 1:
        return 'green'
    else:
        return 'red'

# 2. Aplicamos a função à coluna 'class' e guardamos o resultado na nova coluna 'marker_color'
spacex_df['marker_color'] = spacex_df['class'].apply(assign_marker_color)

# Visualizar as últimas 10 linhas para confirmar a nova coluna
spacex_df.tail(10)
```

*TODO:* For each launch result in `spacex_df` data frame, add a `folium.Marker` to `marker_cluster`


```python
# 1. Primeiro, adicionamos o objeto marker_cluster ao mapa
site_map.add_child(marker_cluster)

# 2. Percorremos cada linha do dataframe spacex_df
for index, record in spacex_df.iterrows():
    # Extraímos a coordenada da linha atual
    coordinate = [record['Lat'], record['Long']]
    
    # Criamos o marcador
    # O 'color' define a cor do pino (geralmente branco para contraste)
    # O 'icon_color' recebe a cor que calculámos no exercício anterior (green/red)
    marker = folium.Marker(
        location=coordinate,
        icon=folium.Icon(color='white', icon_color=record['marker_color']),
        popup=f"Site: {record['Launch Site']}\nStatus: {'Success' if record['class']==1 else 'Failure'}"
    )
    
    # 3. ADICIONAR o marcador ao cluster (e não diretamente ao mapa)
    marker_cluster.add_child(marker)

# Exibir o mapa final
site_map
```

Next, we need to explore and analyze the proximities of launch sites.


Let's first add a `MousePosition` on the map to get coordinate for a mouse over a point on the map. As such, while you are exploring the map, you can easily find the coordinates of any points of interests (such as railway)


```python
# Add Mouse Position to get the coordinate (Lat, Long) for a mouse over on the map
formatter = "function(num) {return L.Util.formatNum(num, 5);};"
mouse_position = MousePosition(
    position='topright',
    separator=' Long: ',
    empty_string='NaN',
    lng_first=False,
    num_digits=20,
    prefix='Lat:',
    lat_formatter=formatter,
    lng_formatter=formatter,
)

site_map.add_child(mouse_position)
site_map
```

Now zoom in to a launch site and explore its proximity to see if you can easily find any railway, highway, coastline, etc. Move your mouse to these points and mark down their coordinates (shown on the top-left) in order to the distance to the launch site.

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closet coastline
# e.g.,: Lat: 28.56367  Lon: -80.57163
# distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
# 1. Definir as coordenadas do local de lançamento e da costa próxima
# Exemplo para o site: CCAFS SLC-40
launch_site_lat = 28.563197
launch_site_lon = -80.576820

# Coordenada aproximada da costa (obtida via MousePosition)
coastline_lat = 28.56367
coastline_lon = -80.56763

# 2. Calcular a distância usando a função fornecida
distance_coastline = calculate_distance(launch_site_lat, launch_site_lon, coastline_lat, coastline_lon)
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property 
# for example
# distance_marker = folium.Marker(
#    coordinate,
#    icon=DivIcon(
#        icon_size=(20,20),
#        icon_anchor=(0,0),
#        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance),
#        )
#    )
# 3. Criar o marcador para o ponto da costa
coordinate = [coastline_lat, coastline_lon]
distance_marker = folium.Marker(
    coordinate,
    icon=DivIcon(
        icon_size=(20,20),
        icon_anchor=(0,0),
        html='<div style="font-size: 12; color:#d35400;"><b>%s</b></div>' % "{:10.2f} KM".format(distance_coastline),
    )
)

# 4. Adicionar o marcador ao mapa
site_map.add_child(distance_marker)

# Exibir o mapa atualizado
site_map
```

*TODO:* Draw a `PolyLine` between a launch site to the selected coastline point


```python
# Create a `folium.PolyLine` object using the coastline coordinates and launch site coordinate
# lines=folium.PolyLine(locations=coordinates, weight=1)
# 1. Definimos a lista de coordenadas (o ponto de partida e o ponto de destino)
# Usando as variáveis que definimos no passo anterior
coordinates = [
    [launch_site_lat, launch_site_lon], 
    [coastline_lat, coastline_lon]
]

# 2. Criamos o objeto PolyLine
# O parâmetro 'weight' define a espessura da linha
lines = folium.PolyLine(
    locations=coordinates, 
    weight=2,        # Espessura da linha
    color='blue',    # Cor da linha para destacar sobre o mapa
    opacity=0.8      # Opacidade (transparência)
)

# 3. Adicionamos a linha ao mapa
site_map.add_child(lines)

# Exibir o mapa com a linha desenhada
site_map
```

```python
# Create a marker with distance to a closest city, railway, highway, etc.
# Draw a line between the marker to the launch site
# 1. Definição dos pontos de interesse (ajuste as coordenadas usando o MousePosition)
launch_site_coords = [28.563197, -80.576820]

proximities = {
    "Railway": [28.57206, -80.58525],
    "Highway": [28.5632, -80.57079],
    "City (Titusville)": [28.6115, -80.8077]
}

# 2. Ciclo para calcular, marcar e desenhar as linhas
for name, poi_coords in proximities.items():
    # Calcular distância
    dist = calculate_distance(launch_site_coords[0], launch_site_coords[1], poi_coords[0], poi_coords[1])
    
    # Criar o Marcador de Distância
    dist_marker = folium.Marker(
        poi_coords,
        icon=DivIcon(
            icon_size=(20,20),
            icon_anchor=(0,0),
            html='<div style="font-size: 12; color:#2980b9;"><b>%s</b></div>' % "{:10.2f} KM".format(dist),
        ),
        popup=name
    )
    site_map.add_child(dist_marker)
    
    # Criar a Linha (PolyLine)
    line = folium.PolyLine(
        locations=[launch_site_coords, poi_coords],
        weight=2,
        color='green' if "City" not in name else 'red', # Vermelho para cidades (alerta de segurança)
        opacity=0.6
    )
    site_map.add_child(line)

# Exibir o mapa final consolidado
site_map
```
